# Flow Lab — Notebook 03: incompressible solver

This notebook implements two new components: semi-Lagrangian advection and pressure projection. The objective is to transport a field and reduce velocity divergence while recording numerical uncertainty.

```python
import sys, pathlib
sys.path.append(str(pathlib.Path('..').resolve() / 'src'))
import numpy as np
from incompressible import advect, pressure_project, poiseuille_profile
```

## Hypothesis
A pressure projection step should reduce the mean divergence of a velocity field. Semi-Lagrangian backtracing should transport a localized tracer in the direction of velocity, with some numerical smoothing.

```python
h,w=32,64
u=np.ones((h,w)); v=np.zeros((h,w))
tracer=np.zeros((h,w)); tracer[h//2,w//4]=1
transported=advect(tracer,u,v,dt=1.0)
projected_u,projected_v,mean_divergence=pressure_project(u,v)
print({'mean_divergence_after_projection':mean_divergence,'tracer_mass':float(transported.sum())})
```

## Analytical benchmark

For steady planar channel flow, the reference profile is parabolic: `u(y)=G y(H−y)/(2ν)`. A production benchmark will compare the numerical steady state against this profile using L1 and L2 errors.

```python
reference=poiseuille_profile(height=32,pressure_gradient=1.0,viscosity=1.0)
print({'wall_velocity':float(reference[0]),'center_velocity':float(reference[len(reference)//2]),'profile_length':len(reference)})
```

## Interpretation

Observation must be taken from the measured output. Projection is successful only if divergence decreases relative to the pre-projection field. Advection is stable but intentionally diffusive. These are numerical properties, not proof of a general Navier–Stokes solution.
